```python
from src.keithley.simulator import Keithley2450Simulator
from src.keithley.load_model import OhmicLoad
from src.keithley.measurement_controller import MeasurementController

import matplotlib.pyplot as plt 

device = Keithley2450Simulator(load_model=OhmicLoad(1000))
controller = MeasurementController(device)

data = controller.voltage_sweep(0, 5, 0.5)

len(data), data[:2]

voltages = [d["set_voltage"] for d in data]
currents = [d["measured_current"] for d in data]

plt.plot(voltages, currents)
plt.xlabel("Voltage (V)")
plt.ylabel("Current (A)")
plt.title("I-V Sweep")
plt.grid(True)
plt.show()
```
