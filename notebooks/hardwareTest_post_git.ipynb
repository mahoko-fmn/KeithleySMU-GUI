```python
from keithley.hardware_socket import Keithley2450Hardware

smu= Keithley2450Hardware(
        "TCPIP0::192.168.50.200::5025::SOCKET"
     )

smu.inst.write(":SOUR:FUNC VOLT")
smu.inst.write(":SOUR:VOLT 1")
smu.inst.write(':SENS:FUNC "CURR"')
smu.inst.write(":OUTP ON")

smu.inst.write("READ?")
print(smu.inst.read())

smu.inst.write(":OUTP OFF")
smu.close()
```
