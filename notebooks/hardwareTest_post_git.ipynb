```python
# initial test, without source mode switch
from keithley.hardware_socket import Keithley2450Hardware

for i in range(3):
    smu= Keithley2450Hardware(
        "TCPIP0::192.168.50.200::5025::SOCKET"
    )

    print(smu.identify())
    smu.close()
```

```python
from keithley.hardware_socket import Keithley2450Hardware

smu= Keithley2450Hardware(
        "TCPIP0::192.168.50.200::5025::SOCKET"
     )

smu.inst.write(":SOUR:FUNC VOLT")
smu.inst.write(":SOUR:VOLT 1")
smu.inst.write(':SENS:FUNC "CURR"')
smu.inst.write(":OUTP ON")

smu.inst.write("READ?")
print(smu.inst.read())

smu.inst.write(":OUTP OFF")
smu.close()
```

```python
# identiy test with the source switch mode addition 
from keithley.hardware_socket import Keithley2450Hardware

device = Keithley2450Hardware("TCPIP0::192.168.50.200::5025::SOCKET")

device.connect()
print(device.identify())
device.disconnect()
```

```python
# testing illegal setpoint
from keithley.hardware_socket import Keithley2450Hardware

device = Keithley2450Hardware("TCPIP0::192.168.50.200::5025::SOCKET")

device.connect()
device.set_source_mode("voltage")
device.set_voltage_setpoint(1)
device.output_on()

print (device.measure())
device.output_off()
```
